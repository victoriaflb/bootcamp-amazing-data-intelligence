### Tratar inconsistências

```python
import pandas as pd

# 1. Tratamento de espaços em branco extras em colunas de texto
colunas_texto = df.select_dtypes(include=['object', 'string']).columns
for col in colunas_texto:
    df[col] = df[col].astype(str).str.strip()

# 2. Preenchimento de valores nulos com padrões mapeados no Pentaho
valores_padrao = {
    'unidade_da_federacao': 'Não informado',
    'sexo': 'Não informado',
    'idade': 'Não informado',
    'idade_do_morador_na_data_de_referencia': 'Não informado',
    'cor_ou_raca': 'Não informado',
    'tipo_do_domicilio': 'Não informado',
    'condicao_ocupacao': 'Não informado',
    'condicao_de_ocupacao_do_domicilio': 'Não informado',
    'ano_de_referencia': '0',
    'condicao_de_atividade': 'Não informado',
    'horas_trabalhadas_semana': 'Não informado',
    'praticou_esporte_365_dias': 'Não informado'
}
df.fillna(value=valores_padrao, inplace=True)

# 3. Mapeamento da Região a partir da Unidade da Federação (UF)
mapeamento_regiao = {
    # Norte
    'Rondônia': 'Norte', 'Acre': 'Norte', 'Amazonas': 'Norte', 'Roraima': 'Norte',
    'Pará': 'Norte', 'Amapá': 'Norte', 'Tocantins': 'Norte',
    # Nordeste
    'Maranhão': 'Nordeste', 'Piauí': 'Nordeste', 'Ceará': 'Nordeste', 
    'Rio Grande do Norte': 'Nordeste', 'Paraíba': 'Nordeste', 'Pernambuco': 'Nordeste', 
    'Alagoas': 'Nordeste', 'Sergipe': 'Nordeste', 'Bahia': 'Nordeste',
    # Sudeste
    'Minas Gerais': 'Sudeste', 'Espírito Santo': 'Sudeste', 
    'Rio de Janeiro': 'Sudeste', 'São Paulo': 'Sudeste',
    # Sul
    'Paraná': 'Sul', 'Santa Catarina': 'Sul', 'Rio Grande do Sul': 'Sul',
    # Centro-Oeste
    'Mato Grosso do Sul': 'Centro-Oeste', 'Mato Grosso': 'Centro-Oeste', 
    'Goiás': 'Centro-Oeste', 'Distrito Federal': 'Centro-Oeste'
}

uf_col = 'unidade_da_federacao'
if uf_col in df.columns:
    df['regiao'] = df[uf_col].map(mapeamento_regiao).fillna('Não informado')

# 4. Conversão de variáveis quantitativas/numéricas
cols_numericas = [
    'idade', 
    'idade_do_morador_na_data_de_referencia',
    'horas_trabalhadas_semana', 
    'horas_habitualmente_trabalhadas_por_semana_em_todos_os_trabalhos',
    'meses_praticou_esporte_frequencia',
    'tempo_dedicado_esporte_dia',
    'peso_pessoa',
    'peso_da_pessoa',
    'peso_morador_esporte'
]

for col in cols_numericas:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')

print("Tratamento de inconsistências concluído com sucesso!")
```

### Tratar Distribuições

```python
import pandas as pd

print("=== 1. RESUMO ESTATÍSTICO DE VARIÁVEIS NUMÉRICAS ===")
# Seleciona colunas numéricas disponíveis no DataFrame
cols_num = df.select_dtypes(include=['int64', 'float64']).columns
display(df[cols_num].describe().T[['count', 'mean', 'std', 'min', '50%', 'max']])

print("\n=== 2. DISTRIBUIÇÃO DAS PRINCIPAIS CATEGORIAS ===")
cols_cat_chave = ['sexo', 'regiao', 'cor_ou_raca', 'praticou_esporte_365_dias']

for col in cols_cat_chave:
    if col in df.columns:
        print(f"\n--- Distribuição: {col} ---")
        freq = df[col].value_counts(dropna=False)
        perc = df[col].value_counts(normalize=True, dropna=False) * 100
        df_dist = pd.DataFrame({'Frequência': freq, 'Percentual (%)': perc.round(2)})
        display(df_dist)

print("\n=== 3. AMPLIUDE E ASSIMETRIA DAS VARIÁVEIS QUANTITATIVAS ===")

# Mapeia os nomes originais da PNAD para os cálculos de assimetria
cols_para_analisar = [
    'idade_do_morador_na_data_de_referencia',
    'idade',
    'horas_habitualmente_trabalhadas_por_semana_em_todos_os_trabalhos',
    'horas_trabalhadas_semana',
    'tempo_dedicado_esporte_dia',
    'meses_praticou_esporte_frequencia',
    'peso_pessoa',
    'peso_morador_esporte'
]

# Executa o cálculo para as colunas presentes no DataFrame
for col in cols_para_analisar:
    if col in df.columns and df[col].notna().sum() > 0:
        skew = df[col].skew()
        classificacao = (
            "Assimétrica à direita" if skew > 0.5 
            else "Assimétrica à esquerda" if skew < -0.5 
            else "Aproximadamente simétrica"
        )
        print(f"Assimetria (Skewness) de '{col}': {skew:.2f} ({classificacao})")
```

### Outliners

```python
import pandas as pd

# Variáveis numéricas a analisar
cols_outliers = [
    'idade_do_morador_na_data_de_referencia',
    'horas_habitualmente_trabalhadas_por_semana_em_todos_os_trabalhos',
    'tempo_dedicado_esporte_dia',
    'meses_praticou_esporte_frequencia'
]

print("=== IDENTIFICAÇÃO DE OUTLIERS (MÉTODO IQR) ===")

resultados_outliers = []

for col in cols_outliers:
    if col in df.columns and df[col].notna().sum() > 0:
        dados = df[col].dropna()
        
        q1 = dados.quantile(0.25)
        q3 = dados.quantile(0.75)
        iqr = q3 - q1
        
        limite_inferior = q1 - 1.5 * iqr
        limite_superior = q3 + 1.5 * iqr
        
        outliers = dados[(dados < limite_inferior) | (dados > limite_superior)]
        qtd_outliers = len(outliers)
        perc_outliers = (qtd_outliers / len(dados)) * 100
        
        resultados_outliers.append({
            'Variável': col,
            'Q1 (25%)': round(q1, 2),
            'Q3 (75%)': round(q3, 2),
            'IQR': round(iqr, 2),
            'Limite Inferior': round(limite_inferior, 2),
            'Limite Superior': round(limite_superior, 2),
            'Qtd Outliers': qtd_outliers,
            '% Outliers': f"{perc_outliers:.2f}%"
        })

df_outliers = pd.DataFrame(resultados_outliers)
display(df_outliers)
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")

# 1. Identifica a coluna correta de Prática de Esporte
col_esporte = None
candidatos_esporte = [
    'praticou_esporte_365_dias',
    'no_periodo_de_referencia_de_365_dias_praticou_algum_esporte_no_tempo_livre_fora_do_horario_de_trabalho_e_de_educacao_fisica_na_escola'
]

for c in candidatos_esporte:
    if c in df.columns:
        col_esporte = c
        break

# 2. Identifica a coluna correta de Idade
col_idade = None
candidatos_idade = ['idade_do_morador_na_data_de_referencia', 'idade']

for c in candidatos_idade:
    if c in df.columns:
        col_idade = c
        break

# --- TABELAS CRUZADAS (RELAÇÕES) ---

if 'sexo' in df.columns and col_esporte:
    print("=== PRÁTICA DE ESPORTE POR SEXO (%) ===")
    display(pd.crosstab(df['sexo'], df[col_esporte], normalize='index').round(2) * 100)

if 'regiao' in df.columns and col_esporte:
    print("\n=== PRÁTICA DE ESPORTE POR REGIÃO (%) ===")
    display(pd.crosstab(df['regiao'], df[col_esporte], normalize='index').round(2) * 100)

print("\n=== 5. GERANDO GRÁFICOS EXPLORATÓRIOS ===")

df_plot = df.copy()
if col_idade:
    df_plot[col_idade] = pd.to_numeric(df_plot[col_idade], errors='coerce')

# Gráfico 1: Curva de Densidade de Idade por Prática de Esporte (KDE Plot)
if col_esporte and col_idade:
    plt.figure(figsize=(10, 5))
    sns.kdeplot(
        data=df_plot, 
        x=col_idade, 
        hue=col_esporte, 
        common_norm=False, 
        fill=True, 
        alpha=0.4, 
        palette='Set1'
    )
    plt.title('Distribuição de Densidade de Idade por Prática de Esporte', fontsize=12, fontweight='bold')
    plt.xlabel('Idade (Anos)')
    plt.ylabel('Densidade de Pessoas')
    plt.xlim(0, 100)
    plt.tight_layout()
    plt.show()

# Gráfico 2: Proporção de Prática de Esporte por Faixa Etária (%)
if col_esporte and col_idade:
    bins = [0, 14, 29, 49, 64, 120]
    labels = ['0-14 anos', '15-29 anos', '30-49 anos', '50-64 anos', '65+ anos']
    df_plot['faixa_etaria'] = pd.cut(df_plot[col_idade], bins=bins, labels=labels)

    cross_idade = pd.crosstab(df_plot['faixa_etaria'], df_plot[col_esporte], normalize='index') * 100

    cross_idade.plot(kind='bar', figsize=(10, 5), color=['#2ca02c', '#d62728'], edgecolor='black', rot=0)
    plt.title('Proporção de Prática de Esporte por Faixa Etária (%)', fontsize=12, fontweight='bold')
    plt.xlabel('Faixa Etária')
    plt.ylabel('Percentual (%)')
    plt.legend(title='Praticou Esporte?')
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()

# Gráfico 3: Contagem de Prática de Esporte por Região
if 'regiao' in df.columns and col_esporte:
    plt.figure(figsize=(10, 5))
    sns.countplot(data=df_plot, x='regiao', hue=col_esporte, palette='Set2')
    plt.title('Volume de Prática de Esporte por Região', fontsize=12, fontweight='bold')
    plt.xlabel('Região')
    plt.ylabel('Quantidade de Pessoas')
    plt.legend(title='Praticou Esporte?')
    plt.tight_layout()
    plt.show()
```
