### Tratamento de inconsistências

```python
import pandas as pd

print("\n=== TRATAMENTO DE INCONSISTÊNCIAS ===")

# 1. Identificar colunas textuais
colunas_texto = df.select_dtypes(
    include=["object", "string"]
).columns

# 2. Detectar espaços extras
def tem_espaco_sobrando(col):
    preenchidos = col.dropna().astype(str)
    return (preenchidos.str.strip() != preenchidos).sum()

espacos_antes = df[colunas_texto].apply(tem_espaco_sobrando)

espacos_detectados = espacos_antes[espacos_antes > 0]

print("\nColunas com espaços extras:")
print(espacos_detectados)

# 3. Remover espaços das extremidades
for col in colunas_texto:
    df[col] = df[col].str.strip()

# 4. Padronizar caixa apenas nas variáveis categóricas escolhidas
colunas_categoricas = [
    col for col in ["sexo", "cor_ou_raca"]
    if col in df.columns
]

for col in colunas_categoricas:
    df[col] = df[col].str.lower()

# 5. Verificar valores numéricos não negativos
colunas_nao_negativas = [
    "idade_do_morador_na_data_de_referencia",
    "peso_da_pessoa",
    "total_de_moradores",
    "numero_de_comodos_do_domicilio"
]

for col in colunas_nao_negativas:
    if col in df.columns:

        valores = pd.to_numeric(
            df[col],
            errors="coerce"
        )

        negativos = (valores < 0).sum()

        nao_convertiveis = (
            df[col].notna() & valores.isna()
        ).sum()

        print(
            f"\n{col}:"
            f"\n  Valores negativos: {negativos}"
            f"\n  Valores não convertíveis: {nao_convertiveis}"
        )

# 6. Verificar faixa etária plausível
idade_col = "idade_do_morador_na_data_de_referencia"

if idade_col in df.columns:

    idade = pd.to_numeric(
        df[idade_col],
        errors="coerce"
    )

    fora_da_faixa = (
        (idade < 0) | (idade > 110)
    )

    print(
        "\nRegistros com idade fora da faixa definida (0–110):",
        fora_da_faixa.sum()
    )

# 7. Conferir categorias
for col in colunas_categoricas:
    print(f"\nCategorias encontradas em '{col}':")
    print(df[col].value_counts(dropna=False))

# 8. Verificar valores ausentes
colunas_verificacao = [
    col for col in [
        "sexo",
        "cor_ou_raca",
        idade_col
    ]
    if col in df.columns
]

print("\nValores ausentes nas variáveis selecionadas:")
print(df[colunas_verificacao].isna().sum())
```

```python
import pandas as pd

# Base filtrada apenas com as pessoas que efetivamente responderam ao Suplemento de Esporte
col_pratica = 'no_periodo_de_referencia_de_365_dias_praticou_algum_esporte_no_tempo_livre_fora_do_horario_de_trabalho_e_de_educacao_fisica_na_escola'

if 'chave_pessoa' in df.columns:
    df_pessoa = df.drop_duplicates(subset='chave_pessoa').copy()
else:
    df_pessoa = df.copy()

# ==============================================================================
# SEÇÃO 4: FREQUÊNCIA DA PRÁTICA E TEMPO DEDICADO
# ==============================================================================
print("=== 4. FREQUÊNCIA E TEMPO DEDICADO À ATIVIDADE FÍSICA / ESPORTE ===")

cols_freq_esporte = [
    'frequencia_com_que_praticou_esporte',
    'em_dia_que_costumava_praticar_esporte_o_tempo_que_normalmente_dedicava_era_de',
    'em_dia_que_costumava_praticar_atividade_fisica_o_tempo_que_normalmente_dedicava_era_de',
    'quantidade_de_meses_em_que_praticou_esporte'
]

cols_freq_reais = [c for c in cols_freq_esporte if c in df_pessoa.columns]

if cols_freq_reais:
    for col_f in cols_freq_reais:
        # Filtra apenas quem respondeu a variável (removendo os nulos do filtro do suplemento)
        serie_valida = df_pessoa[col_f].dropna()
        if len(serie_valida) > 0:
            distrib = serie_valida.value_counts()
            perc = serie_valida.value_counts(normalize=True) * 100
            df_f = pd.DataFrame({'Frequência Absoluta': distrib, 'Percentual (%)': perc.round(2)})
            print(f"\n--- Tabela: {col_f} ---")
            display(df_f.head(10))
else:
    # Procura colunas com 'tempo' ou 'meses' relacionadas a esporte
    cols_alt = [c for c in df_pessoa.columns if 'esport' in c.lower() and any(p in c.lower() for p in ['tempo', 'meses', 'frequen', 'dia'])]
    for col_a in cols_alt[:3]:
        serie_v = df_pessoa[col_a].dropna()
        if len(serie_v) > 0:
            distrib = serie_v.value_counts()
            perc = serie_v.value_counts(normalize=True) * 100
            print(f"\n--- Tabela: {col_a} ---")
            display(pd.DataFrame({'Frequência Absoluta': distrib, 'Percentual (%)': perc.round(2)}).head(10))

# ==============================================================================
# SEÇÃO 5: MOTIVOS DE NÃO PRÁTICA / ABANDONO
# ==============================================================================
print("\n=== 5. MOTIVOS DE NÃO PRÁTICA / PARADA DO ESPORTE ===")

cols_motivos_esporte = [
    'qual_o_principal_motivo_de_nao_ter_praticado_esporte',
    'qual_o_principal_motivo_de_nao_ter_praticado_atividade_fisica',
    'motivo_pelo_qual_parou_de_praticar_esporte',
    'idade_que_tinha_quando_deixou_de_praticar_esporte'
]

cols_motivos_reais = [c for c in cols_motivos_esporte if c in df_pessoa.columns]

if cols_motivos_reais:
    for col_m in cols_motivos_reais:
        serie_m = df_pessoa[col_m].dropna()
        if len(serie_m) > 0:
            distrib_m = serie_m.value_counts()
            perc_m = serie_m.value_counts(normalize=True) * 100
            df_m = pd.DataFrame({'Frequência Absoluta': distrib_m, 'Percentual (%)': perc_m.round(2)})
            print(f"\n--- Tabela: {col_m} ---")
            display(df_m.head(10))
else:
    # Procura colunas com 'motivo' ou 'parou'
    cols_m_alt = [c for c in df_pessoa.columns if any(p in c.lower() for p in ['motivo', 'parou', 'deixou']) and any(p in c.lower() for p in ['esport', 'atividad', 'prat'])]
    for col_ma in cols_m_alt[:3]:
        serie_mv = df_pessoa[col_ma].dropna()
        if len(serie_mv) > 0:
            distrib_m = serie_mv.value_counts()
            perc_m = serie_mv.value_counts(normalize=True) * 100
            print(f"\n--- Tabela: {col_ma} ---")
            display(pd.DataFrame({'Frequência Absoluta': distrib_m, 'Percentual (%)': perc_m.round(2)}).head(10))
```

### Outliners

```python
import pandas as pd
import numpy as np


# Mapeamento dinâmico das variáveis numéricas do tema
col_idade = 'idade_do_morador_na_data_de_referencia' if 'idade_do_morador_na_data_de_referencia' in df_pessoa.columns else 'idade'
col_tempo = 'em_dia_que_costumava_praticar_atividade_fisica_o_tempo_que_normalmente_dedicava_era_de'
col_meses = 'quantidade_de_meses_em_que_praticou_esporte'

cols_para_outliers = [c for c in [col_idade, col_tempo, col_meses] if c in df_pessoa.columns]

print("=== IDENTIFICAÇÃO DE OUTLIERS (MÉTODO IQR COM CONTEXTO DE BI) ===")

resultados_outliers = []

for col in cols_para_outliers:
    # Garante conversão numérica tratando vírgula
    s_limpa = df_pessoa[col].astype(str).str.replace(',', '.', regex=False).str.strip()
    dados = pd.to_numeric(s_limpa, errors='coerce').dropna()
    
    if len(dados) > 0:
        q1 = dados.quantile(0.25)
        q3 = dados.quantile(0.75)
        iqr = q3 - q1
        
        limite_inferior = q1 - 1.5 * iqr
        limite_superior = q3 + 1.5 * iqr
        
        outliers = dados[(dados < limite_inferior) | (dados > limite_superior)]
        qtd_outliers = len(outliers)
        perc_outliers = (qtd_outliers / len(dados)) * 100
        
        # Avaliação conceitual de negócio
        if col == col_idade:
            diagnostico = "Dado biológico real (idosos longevos). NÃO REMOVER."
        elif dados.max() <= 12:
            diagnostico = "Escala de códigos/meses. Avaliar por categorias."
        else:
            diagnostico = "Investigar valores atípicos extemporâneos."
            
        resultados_outliers.append({
            'Variável': col,
            'n (Válidos)': len(dados),
            'Q1 (25%)': round(q1, 2),
            'Q3 (75%)': round(q3, 2),
            'IQR': round(iqr, 2),
            'Lim. Inf': round(limite_inferior, 2),
            'Lim. Sup': round(limite_superior, 2),
            'Qtd Outliers': qtd_outliers,
            '% Outliers': f"{perc_outliers:.2f}%",
            'Recomendação BI': diagnostico
        })

df_outliers_resumo = pd.DataFrame(resultados_outliers)
display(df_outliers_resumo)
```

```python
import pandas as pd

import seaborn as sns

import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")

# 1. Identifica a coluna correta de Prática de Esporte

col_esporte = None

candidatos_esporte = [

    'praticou_esporte_365_dias',

    'no_periodo_de_referencia_de_365_dias_praticou_algum_esporte_no_tempo_livre_fora_do_horario_de_trabalho_e_de_educacao_fisica_na_escola'

]

for c in candidatos_esporte:

    if c in df.columns:

        col_esporte = c

        break

# 2. Identifica a coluna correta de Idade

col_idade = None

candidatos_idade = ['idade_do_morador_na_data_de_referencia', 'idade']

for c in candidatos_idade:

    if c in df.columns:

        col_idade = c

        break

# --- TABELAS CRUZADAS (RELAÇÕES) ---

if 'sexo' in df.columns and col_esporte:

    print("=== PRÁTICA DE ESPORTE POR SEXO (%) ===")

    display(pd.crosstab(df['sexo'], df[col_esporte], normalize='index').round(2) * 100)

if 'regiao' in df.columns and col_esporte:

    print("\n=== PRÁTICA DE ESPORTE POR REGIÃO (%) ===")

    display(pd.crosstab(df['regiao'], df[col_esporte], normalize='index').round(2) * 100)

print("\n=== 5. GERANDO GRÁFICOS EXPLORATÓRIOS ===")

df_plot = df.copy()

if col_idade and col_esporte:

    # Converter idade para formato numérico

    df_plot[col_idade] = pd.to_numeric(

        df_plot[col_idade],

        errors="coerce"

    )

    # Remover registros sem idade ou resposta de esporte

    df_plot = df_plot.dropna(

        subset=[col_idade, col_esporte]

    ).copy()

    # Manter idades válidas

    df_plot = df_plot[

        df_plot[col_idade].between(0, 100)

    ].copy()

    # Criar faixas etárias

    limites = [0, 18, 30, 40, 50, 60, 101]

    rotulos = [

        "0–17",

        "18–29",

        "30–39",

        "40–49",

        "50–59",

        "60+"

    ]

    df_plot["faixa_etaria"] = pd.cut(

        df_plot[col_idade],

        bins=limites,

        labels=rotulos,

        right=False

    )

    # Conferir as categorias existentes na variável

    print("Respostas encontradas:")

    print(df_plot[col_esporte].value_counts(dropna=False))

    # ATENÇÃO:

    # Confirme o valor que representa uma resposta afirmativa

    # antes de calcular as proporções.

    valor_sim = "Sim"

    df_plot["praticou_esporte"] = (

        df_plot[col_esporte]

        .astype(str)

        .str.strip()

        .str.casefold()

        .eq(valor_sim.casefold())

    )

    # Calcular proporção por faixa etária

    resultado = (

        df_plot.groupby(

            "faixa_etaria",

            observed=False

        )["praticou_esporte"]

        .mean()

        .mul(100)

        .reset_index(name="proporcao")

    )

    # Gráfico

    plt.figure(figsize=(10, 5))

    ax = sns.barplot(

        data=resultado,

        x="faixa_etaria",

        y="proporcao"

    )

    ax.bar_label(

        ax.containers[0],

        fmt="%.1f%%",

        padding=3

    )

    plt.title(

        "Proporção de praticantes de esporte por faixa etária",

        fontsize=13,

        fontweight="bold"

    )

    plt.xlabel("Faixa etária (anos)")

    plt.ylabel("Praticantes de esporte (%)")

    plt.ylim(0, 100)

    plt.tight_layout()

    plt.show()

else:

    print("Não foi possível identificar as colunas de idade e esporte.")

# Gráfico 2: Proporção de Prática de Esporte por Faixa Etária (%)

if col_esporte and col_idade:

    bins = [0, 14, 29, 49, 64, 120]

    labels = ['0-14 anos', '15-29 anos', '30-49 anos', '50-64 anos', '65+ anos']

    df_plot['faixa_etaria'] = pd.cut(df_plot[col_idade], bins=bins, labels=labels)

    cross_idade = pd.crosstab(df_plot['faixa_etaria'], df_plot[col_esporte], normalize='index') * 100

    cross_idade.plot(kind='bar', figsize=(10, 5), color=['#2ca02c', '#d62728'], edgecolor='black', rot=0)

    plt.title('Proporção de Prática de Esporte por Faixa Etária (%)', fontsize=12, fontweight='bold')

    plt.xlabel('Faixa Etária')

    plt.ylabel('Percentual (%)')

    plt.legend(title='Praticou Esporte?')

    plt.grid(axis='y', linestyle='--', alpha=0.7)

    plt.tight_layout()

    plt.show()

# Gráfico 3: Contagem de Prática de Esporte por Região

if 'regiao' in df.columns and col_esporte:

    plt.figure(figsize=(10, 5))

    sns.countplot(data=df_plot, x='regiao', hue=col_esporte, palette='Set2')

    plt.title('Volume de Prática de Esporte por Região', fontsize=12, fontweight='bold')

    plt.xlabel('Região')

    plt.ylabel('Quantidade de Pessoas')

    plt.legend(title='Praticou Esporte?')

    plt.tight_layout()

    plt.show()
```
