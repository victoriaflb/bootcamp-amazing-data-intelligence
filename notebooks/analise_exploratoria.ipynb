### Identificar tipos de dados incorretos

```python
import pandas as pd

# 1. Identifica AUTOMATICAMENTE todas as colunas do df que possuem números ou podem ser convertidas
candidatas_numericas = []

for col in df.columns:
    # Ignora colunas de identificadores/chaves
    if not any(id_kw in col.lower() for id_kw in ['chave', 'id_', 'cd_', 'cod_']):
        # Limpa vírgulas e espaços temporariamente para testar
        amostra_limpa = df[col].dropna().astype(str).str.replace(',', '.', regex=False).str.strip()
        # Converte a amostra para teste
        test_numeric = pd.to_numeric(amostra_limpa, errors='coerce')
        
        # Se contiver números válidos E mais de 5 valores distintos (evita categóricas puras como '1', '2' para Sexo)
        if test_numeric.notna().sum() > 0 and test_numeric.nunique() > 5:
            candidatas_numericas.append(col)

print(f"=== FORAM ENCONTRADAS {len(candidatas_numericas)} COLUNAS NUMÉRICAS REAIS EM SEU DATASET ===")

# 2. Registra nulos antes do tratamento
nulos_antes = df[candidatas_numericas].isna().sum()

# 3. Tratamento prévio de vírgula decimal e conversão segura
for col in candidatas_numericas:
    if df[col].dtype == 'object':
        df[col] = df[col].astype(str).str.replace(',', '.', regex=False).str.strip()

df[candidatas_numericas] = df[candidatas_numericas].apply(pd.to_numeric, errors='coerce')

# 4. Registra nulos depois da conversão
nulos_depois = df[candidatas_numericas].isna().sum()
novos_nans = nulos_depois - nulos_antes

# 5. Tabela Final de Diagnóstico de Colunas Numéricas
df_diagnostico_conversao = pd.DataFrame({
    'Tipo de Dado': df[candidatas_numericas].dtypes,
    'Registros Válidos (Números)': df[candidatas_numericas].notna().sum(),
    'Novos NaNs Gerados (Texto/Inválidos)': novos_nans
})

display(df_diagnostico_conversao.sort_values(by='Registros Válidos (Números)', ascending=False))
```

### Tratamento de inconsistências

```python
import pandas as pd

print("\n=== TRATAMENTO DE INCONSISTÊNCIAS ===")

# 1. Identificar colunas textuais
colunas_texto = df.select_dtypes(
    include=["object", "string"]
).columns

# 2. Detectar espaços extras
def tem_espaco_sobrando(col):
    preenchidos = col.dropna().astype(str)
    return (preenchidos.str.strip() != preenchidos).sum()

espacos_antes = df[colunas_texto].apply(tem_espaco_sobrando)

espacos_detectados = espacos_antes[espacos_antes > 0]

print("\nColunas com espaços extras:")
print(espacos_detectados)

# 3. Remover espaços das extremidades
for col in colunas_texto:
    df[col] = df[col].str.strip()

# 4. Padronizar caixa apenas nas variáveis categóricas escolhidas
colunas_categoricas = [
    col for col in ["sexo", "cor_ou_raca"]
    if col in df.columns
]

for col in colunas_categoricas:
    df[col] = df[col].str.lower()

# 5. Verificar valores numéricos não negativos
colunas_nao_negativas = [
    "idade_do_morador_na_data_de_referencia",
    "peso_da_pessoa",
    "total_de_moradores",
    "numero_de_comodos_do_domicilio"
]

for col in colunas_nao_negativas:
    if col in df.columns:

        valores = pd.to_numeric(
            df[col],
            errors="coerce"
        )

        negativos = (valores < 0).sum()

        nao_convertiveis = (
            df[col].notna() & valores.isna()
        ).sum()

        print(
            f"\n{col}:"
            f"\n  Valores negativos: {negativos}"
            f"\n  Valores não convertíveis: {nao_convertiveis}"
        )

# 6. Verificar faixa etária plausível
idade_col = "idade_do_morador_na_data_de_referencia"

if idade_col in df.columns:

    idade = pd.to_numeric(
        df[idade_col],
        errors="coerce"
    )

    fora_da_faixa = (
        (idade < 0) | (idade > 110)
    )

    print(
        "\nRegistros com idade fora da faixa definida (0–110):",
        fora_da_faixa.sum()
    )

# 7. Conferir categorias
for col in colunas_categoricas:
    print(f"\nCategorias encontradas em '{col}':")
    print(df[col].value_counts(dropna=False))

# 8. Verificar valores ausentes
colunas_verificacao = [
    col for col in [
        "sexo",
        "cor_ou_raca",
        idade_col
    ]
    if col in df.columns
]

print("\nValores ausentes nas variáveis selecionadas:")
print(df[colunas_verificacao].isna().sum())
```

### Outliers

```python
import pandas as pd
import numpy as np


# Mapeamento dinâmico das variáveis numéricas do tema
col_idade = 'idade_do_morador_na_data_de_referencia' if 'idade_do_morador_na_data_de_referencia' in df_pessoa.columns else 'idade'
col_tempo = 'em_dia_que_costumava_praticar_atividade_fisica_o_tempo_que_normalmente_dedicava_era_de'
col_meses = 'quantidade_de_meses_em_que_praticou_esporte'

cols_para_outliers = [c for c in [col_idade, col_tempo, col_meses] if c in df_pessoa.columns]

print("=== IDENTIFICAÇÃO DE OUTLIERS (MÉTODO IQR COM CONTEXTO DE BI) ===")

resultados_outliers = []

for col in cols_para_outliers:
    # Garante conversão numérica tratando vírgula
    s_limpa = df_pessoa[col].astype(str).str.replace(',', '.', regex=False).str.strip()
    dados = pd.to_numeric(s_limpa, errors='coerce').dropna()
    
    if len(dados) > 0:
        q1 = dados.quantile(0.25)
        q3 = dados.quantile(0.75)
        iqr = q3 - q1
        
        limite_inferior = q1 - 1.5 * iqr
        limite_superior = q3 + 1.5 * iqr
        
        outliers = dados[(dados < limite_inferior) | (dados > limite_superior)]
        qtd_outliers = len(outliers)
        perc_outliers = (qtd_outliers / len(dados)) * 100
        
        # Avaliação conceitual de negócio
        if col == col_idade:
            diagnostico = "Dado biológico real (idosos longevos). NÃO REMOVER."
        elif dados.max() <= 12:
            diagnostico = "Escala de códigos/meses. Avaliar por categorias."
        else:
            diagnostico = "Investigar valores atípicos extemporâneos."
            
        resultados_outliers.append({
            'Variável': col,
            'n (Válidos)': len(dados),
            'Q1 (25%)': round(q1, 2),
            'Q3 (75%)': round(q3, 2),
            'IQR': round(iqr, 2),
            'Lim. Inf': round(limite_inferior, 2),
            'Lim. Sup': round(limite_superior, 2),
            'Qtd Outliers': qtd_outliers,
            '% Outliers': f"{perc_outliers:.2f}%",
            'Recomendação BI': diagnostico
        })

df_outliers_resumo = pd.DataFrame(resultados_outliers)
display(df_outliers_resumo)
```

### Gráficos exploratórios

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid", font_scale=1.2)

# ---------------------------------------------------------------
# CONFIGURAÇÕES
# ---------------------------------------------------------------
CANDIDATOS_ESPORTE = [
    "praticou_esporte_365_dias",
    "no_periodo_de_referencia_de_365_dias_praticou_algum_esporte_no_tempo_livre_fora_do_horario_de_trabalho_e_de_educacao_fisica_na_escola",
]
CANDIDATOS_IDADE = ["idade_do_morador_na_data_de_referencia", "idade"]

IDADE_MIN, IDADE_MAX = 15, 100  # suplemento de esporte da PNAD 2015: 15 anos ou mais
LIMITES = [15, 30, 45, 60, 101]
ROTULOS = ["15–29", "30–44", "45–59", "60+"]
FIGSIZE = (14, 7)


# ---------------------------------------------------------------
# FUNÇÕES AUXILIARES
# ---------------------------------------------------------------
def achar_coluna(df, candidatos):
    """Retorna o primeiro nome de coluna existente no DataFrame, ou None."""
    return next((c for c in candidatos if c in df.columns), None)


def mostrar_crosstab(df, linha, coluna, titulo):
    """Exibe tabela cruzada em percentual por linha."""
    if linha in df.columns and coluna:
        print(f"\n=== {titulo} (%) ===")
        display((pd.crosstab(df[linha], df[coluna], normalize="index") * 100).round(1))


def preparar_base(df, col_idade, col_esporte):
    """Limpa idade e resposta, filtra faixa válida e cria faixa etária + flag de esporte."""
    base = df.copy()
    base[col_idade] = pd.to_numeric(base[col_idade], errors="coerce")
    base = base.dropna(subset=[col_idade, col_esporte])
    base = base[base[col_idade].between(IDADE_MIN, IDADE_MAX)].copy()

    base["faixa_etaria"] = pd.cut(
        base[col_idade], bins=LIMITES, labels=ROTULOS, right=False
    )
    base["praticou_esporte"] = (
        base[col_esporte].astype(str).str.strip().str.casefold().eq("sim")
    )
    return base


def finalizar(titulo, xlabel, ylabel, legenda=None):
    plt.title(titulo, fontsize=12, fontweight="bold")
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    if legenda:
        plt.legend(title=legenda)
    plt.tight_layout()
    plt.show()


# ---------------------------------------------------------------
# GRÁFICOS
# ---------------------------------------------------------------
def grafico_taxa_por_faixa(base):
    resultado = (
        base.groupby("faixa_etaria", observed=False)["praticou_esporte"]
        .mean()
        .mul(100)
        .reset_index(name="proporcao")
    )

    plt.figure(figsize=FIGSIZE)
    ax = sns.barplot(
        data=resultado,
        x="faixa_etaria",
        y="proporcao",
        hue="faixa_etaria",
        palette="Blues_d",
        width=0.45,
        dodge=False,      # barras largas e centralizadas
        legend=False,
    )
    for container in ax.containers:
        ax.bar_label(container, fmt="%.1f%%", padding=3)

    plt.ylim(0, max(resultado["proporcao"].max() * 1.15, 10))
    finalizar(
        "Proporção de Praticantes de Esporte por Faixa Etária (%)",
        "Faixa Etária (anos)",
        "Praticantes de Esporte (%)",
    )


def grafico_distribuicao_respostas(base, col_esporte):
    cross = pd.crosstab(base["faixa_etaria"], base[col_esporte], normalize="index") * 100

    cross.plot(
        kind="bar",
        figsize=FIGSIZE,
        color=sns.color_palette("Set2", n_colors=cross.shape[1]),  # ajusta ao nº de respostas
        edgecolor="black",
        rot=0,
    )
    plt.grid(axis="y", linestyle="--", alpha=0.7)
    finalizar(
        "Distribuição Relativa de Respostas por Faixa Etária (%)",
        "Faixa Etária",
        "Percentual (%)",
        legenda="Resposta",
    )


def grafico_volume_por_regiao(base, col_esporte):
    df_reg = base.dropna(subset=["regiao", col_esporte])

    plt.figure(figsize=FIGSIZE)
    sns.countplot(data=df_reg, x="regiao", hue=col_esporte, palette="Set2")
    finalizar(
        "Volume Amostral da Prática de Esporte por Região",
        "Região",
        "Quantidade de Pessoas",
        legenda="Resposta",
    )


# ---------------------------------------------------------------
# EXECUÇÃO
# ---------------------------------------------------------------
col_esporte = achar_coluna(df, CANDIDATOS_ESPORTE)
col_idade = achar_coluna(df, CANDIDATOS_IDADE)

if not col_esporte:
    print("⚠️ Coluna de prática de esporte não encontrada.")
else:
    mostrar_crosstab(df, "sexo", col_esporte, "PRÁTICA DE ESPORTE POR SEXO")
    mostrar_crosstab(df, "regiao", col_esporte, "PRÁTICA DE ESPORTE POR REGIÃO")

if col_idade and col_esporte:
    print("\n=== 5. GERANDO GRÁFICOS EXPLORATÓRIOS ===")
    df_plot = preparar_base(df, col_idade, col_esporte)

    grafico_taxa_por_faixa(df_plot)
    grafico_distribuicao_respostas(df_plot, col_esporte)

    if "regiao" in df_plot.columns:
        grafico_volume_por_regiao(df_plot, col_esporte)
else:
    print("⚠️ Coluna de idade ou de esporte não encontrada.")
```
